## 1. Distribución de Maxwell-Boltzmann
Graficamos la distribución de velocidades de Maxwell-Boltzmann para un gas ideal en 3D:
$$ f(v) = 4\pi \left( \frac{m}{2\pi k_B T} \right)^{3/2} v^2 \exp\left(-\frac{m v^2}{2 k_B T}\right) $$

```python
def maxwell_boltzmann(v, T, m=1.0, kB=1.0):
    norm = 4 * np.pi * (m / (2 * np.pi * kB * T))**(1.5)
    return norm * (v**2) * np.exp(-m * (v**2) / (2 * kB * T))

v = np.linspace(0, 5, 500)
for T in [0.5, 1.0, 2.0]:
    plt.plot(v, maxwell_boltzmann(v, T), label=f'T = {T}')

plt.xlabel('Velocidad v')
plt.ylabel('Densidad de probabilidad f(v)')
plt.title('Distribución de Maxwell-Boltzmann')
plt.legend()
plt.show()
```
