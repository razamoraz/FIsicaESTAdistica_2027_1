Este cuaderno introduce el cómputo de alto rendimiento en **Julia** para física estadística y sistemas cuánticos.

```julia
# Julia code snippet for statistical physics
function maxwell_boltzmann(v, T; m=1.0, kB=1.0)
    norm_factor = 4 * pi * (m / (2 * pi * kB * T))^(1.5)
    return norm_factor * (v^2) * exp(-m * (v^2) / (2 * kB * T))
end

println("Maxwell-Boltzmann function defined in Julia.")
```
